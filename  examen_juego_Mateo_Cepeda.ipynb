```python
class Jugador:
    def __init__(self, nombre, simbolo):
        self.nombre = nombre
        self.simbolo = simbolo

    def jugada(self):
        while True:
            try:
                posicion = int(input(f"{self.nombre} ({self.simbolo}), elige una casilla (1-9): "))
                if 1 <= posicion <= 9:
                    return posicion
                print("Error: El número debe estar entre 1 y 9.")
            except ValueError:
                print("Error: Por favor ingresa un número válido.")
```

```python
class TresEnRaya:
    def __init__(self):
        self.tablero = {i: " " for i in range(1, 10)}
    def mostrar_tablero(self):
        t = self.tablero
        print(f"\n {t[1]} | {t[2]} | {t[3]} ")
        print("")
        print(f" {t[4]} | {t[5]} | {t[6]} ")
        print("-----------")
        print(f" {t[7]} | {t[8]} | {t[9]} \n")

    def verificar_ganador(self, s):
        t = self.tablero
        combos = [(1,2,3), (4,5,6), (7,8,9), (1,4,7), (2,5,8), (3,6,9), (1,5,9), (3,5,7)]
        return any(t[a] == t[b] == t[c] == s for a, b, c in combos)

    def jugar(self, j1, j2):
        actual = j1
        for turno in range(9):
            self.mostrar_tablero()

            while True:
                pos = actual.jugada()
                if self.tablero[pos] == " ":
                    self.tablero[pos] = actual.simbolo
                    break
                print("Esa casilla ya está ocupada. Intenta otra.")

            if self.verificar_ganador(actual.simbolo):
                self.mostrar_tablero()
                print(f"¡Felicidades {actual.nombre}! Has ganado.")
                return
            actual = j2 if actual == j1 else j1

        self.mostrar_tablero()
        print("Es un empate.")
```
